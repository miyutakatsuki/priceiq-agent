## 1. Executive Summary
Your team will design, architect, and evaluate an **Agentic AI System** that solves a non-trivial business analytics problem. Moving beyond simple "wrappers," this project requires a deterministic reasoning loop where the AI autonomously manages state and utilizes tools to achieve complex goals. You are responsible for the entire lifecycle: from **Architectural Decision Records (ADRs)** to **FinOps (Cost Optimization)** and **Automated Evaluation Pipelines**.

---

## 2. The "Gatekeeper" Requirement (Phase 1)
To ensure project rigor, your proposal must pass the **Complexity Threshold**. Projects that can be solved with a single prompt or simple RAG will be rejected. Your system must demonstrate:
* **Multi-Step Reasoning:** Requires at least 3 distinct logical steps/loops.
* **Analytical Transformation:** At least one tool must perform a calculation, data pivot, or statistical interpretation (e.g., via `pandas` or `scipy`).
* **Ambiguity Resolution:** The agent must handle vague user intents by probing or selecting the correct tool path.

## 3. Technical Platform Tracks
Teams select **one** track. Both require Python development (Colab/Local).

### Track A: LangGraph (Framework-First)
* **Focus:** State machines, cyclical graphs, and persistence.
* **Requirement:** Implementation of a `StateGraph` with explicit nodes, edges, and conditional logic.

### Track B: Claude Agent SDK (Architecture-First)
* **Focus:** High-precision steering and behavioral alignment.
* **Requirement:** Manual orchestration of the `tool_use` loop. To balance complexity with Track A, this track **requires a Multi-Agent Orchestrator** (e.g., a "Planner" agent and an "Executor" agent).

## 4. Technical Requirements & FinOps

### A. Prompt Version Control (PVC)
You must document the evolution of your system prompts in your final submission. Submit a **Prompt Evolution Log** showing:
1.  **v1 Initial Prompt:** (Describe failure mode).
2.  **v2 Iteration:** (Describe fix, e.g., added XML tagging).
3.  **v3 Final Version:** (Explain why this version stabilized the agent).

### B. FinOps & Guardrails
Since you are managing your own "Burn Rate," your system must include:
* **Hard Limits:** Mandatory `max_iterations` (e.g., 5-10) and token-cap logic to prevent infinite loops and runaway costs.
* **Cost-per-Success Analysis:** In your final report, calculate the cost of a "Success" vs. a "Failure."

### C. Functional Tooling
* **Minimum 3 Tools:** (e.g., SQL execution, Python-based math, External API).
* **Analytic Rigor:** Tools must return structured data that the agent interprets, not just passes through.

## 5. The "Evaluation Pipeline" (Level of Rigor)

Twenty manual test cases are a baseline, not a conclusion, in your final submission. Your evaluation must consist of:

### A. Synthetic Test Generation
For your final submission, use a high-reasoning LLM (e.g., GPT-4o or Claude 3.5 Sonnet) to generate **50 variations** of your 5 core "Seed" test cases. These variations should include:
* **Varying Tones:** (Professional, frustrated, vague).
* **Edge Cases:** (Missing parameters, out-of-bounds dates).

### B. LLM-as-a-Judge
For your final submission, you must build a separate **Evaluator Agent** with a custom rubric to grade your **Worker Agent**.
* **Standard Rubric:** Instruction Adherence, Reasoning Transparency, and Hallucination Check.
* **Custom Rubric:** Define 3 project-specific KPIs (e.g., "Was the financial advice compliant with [specific rule]?").
* **Consistency Score:** Run 10 core test cases **3 times each** and report the variance in the agent's performance.

## 8. Grading Rubric (Final Submission & Presentation)

The following are the rubric categories, criteria and points for the final submission.

**1. Technical Implementation (30 Points)**
*Focus: The "plumbing" and orchestration of the agent.*

| Score | Criteria for "Excellent" (27-30 pts) | Criteria for "Fail" (0-17 pts) |
| :--- | :--- | :--- |
| **Logic & Flow** | Sophisticated use of state (LangGraph) or multi-agent orchestration (Claude). Loops are logical, handle branching, and recover from errors. | Single-prompt "wrapper" or simple linear chain. No recovery logic. |
| **Tooling** | 3+ high-quality tools that return complex data. The agent correctly parses and acts on tool output without manual intervention. | Fewer than 3 tools. Tools often fail or return data the agent cannot process. |
| **Guardrails** | Hard limits on iterations and token spend are coded and functional. System cannot enter an infinite loop. | No safeguards; "unbounded" agent that could drain credits. |


**2. Reasoning & Analytic Rigor (20 Points)**
*Focus: Is the agent "thinking" or just guessing?*

| Score | Criteria for "Excellent" (18-20 pts) | Criteria for "Fail" (0-11 pts) |
| :--- | :--- | :--- |
| **Framework** | Clear evidence of ReAct, CoT, or ToT. Internal monologue is transparent, logical, and adds value to the final result. | No reasoning trace; "Black box" responses with no internal logic shown. |
| **Analysis** | Agent performs true data transformation (e.g., calculating trends, identifying anomalies). It interprets *why* a number matters. | Agent just "regurgitates" data (e.g., "The sales were 500") without context or analysis. |
| **PVC Log** | The **Prompt Version Control** log shows at least 3 distinct iterations with detailed notes on what was fixed. | No log provided, or only one version of the prompt exists. |

**3. Evaluation & Quality Assurance (20 Points)**
*Focus: The "LLM-as-a-Judge" and stress testing.*

| Score | Criteria for "Excellent" (18-20 pts) | Criteria for "Fail" (0-11 pts) |
| :--- | :--- | :--- |
| **Synthetic Tests** | 50+ diverse test cases (adversarial, edge cases, different tones) generated and tracked. | Fewer than 20 test cases, mostly "happy path" (simple queries). |
| **Judge Quality** | The "Judge LLM" has a highly specific rubric. The team reports a **Consistency Score** (3 runs per test case) with honest variance data. | Judge is just told "Grade this." No consistency testing. |
| **Red Teaming** | Clear documentation of how the agent was broken and how the team tried (even if unsuccessfully) to patch it. | No mention of failures or edge cases. |

**4. FinOps & Performance Analysis (15 Points)**
*Focus: Business viability and cost control.*

| Score | Criteria for "Excellent" (13-15 pts) | Criteria for "Fail" (0-8 pts) |
| :--- | :--- | :--- |
| **Cost Metrics** | Detailed breakdown of Cost-per-Success. Comparison of "High Performance" (expensive) vs. "Budget" (cheap) models. | No cost tracking; "Burn rate" ignored. |
| **Latency** | Detailed profiling of where time is spent (Inference vs. Tool Execution). Identify the "bottleneck." | No latency measurements provided. |
| **KPI Targets** | 3+ domain-specific KPIs are tracked and the agent’s performance against them is quantified. | Vague "it works well" statements instead of metrics. |

**5. Documentation & Justification (15 Points)**
*Focus: The "Why" behind the "How."*

| Score | Criteria for "Excellent" (13-15 pts) | Criteria for "Fail" (0-8 pts) |
| :--- | :--- | :--- |
| **ADRs** | **Architecture Decision Records** provide strong evidence for model choice and AI vs. Rule-Based component splits. | Architecture feels random; no justification for why an LLM was used for a task. |
| **Failure Analysis** | Deeply honest post-mortem. Team identifies *why* the model fails (e.g., "Model X lacks spatial reasoning for this task"). | Defensive documentation that hides flaws or failures. |
| **Visuals** | Professional-grade architecture diagram (Mermaid/Lucid) that matches the actual code implementation. | Messy or inaccurate diagrams that don't match the code. |

## 11. Project Examples

To make this project feasible, you may have to use data from:
1.  **Kaggle Datasets** (CSV-based SQL tools).
2.  **Free-Tier APIs** (OpenWeather, NewsAPI, CoinGecko).
3.  **LLM-Generated Synthetic Data** (Perfect for "Mocking" a corporate database).

Here are three feasible, high-impact problem statements with clear data paths:


**Project Suggestion 1: The "Smart E-Commerce Returns Analyst"**
**The Problem:** E-commerce companies lose billions on returns. This agent analyzes why products are being returned and suggests a specific business action (e.g., "Update the size chart" or "Flag this batch for QA").

* **Data Source:** * **Synthetic:** Generate a CSV of 500 "Customer Reviews" and "Return Reason Codes" using an LLM.
    * **SQL Tool:** A SQLite database containing a "Product Catalog" (prices, materials).
* **The Agentic Loop:**
    1.  User asks: "Why are our 'Lunar Running Shoes' being returned so often this month?"
    2.  **Tool 1 (SQL):** Query the return rate for that SKU.
    3.  **Tool 2 (Python/NLTK):** Analyze the text of the reviews for that SKU to find common keywords (e.g., "runs small," "tight toe box").
    4.  **Reasoning:** The agent compares the return rate to a "Normal" threshold and decides if the issue is a design flaw or a sizing description error.
* **Why it works:** You can generate the "bad" data yourself to ensure the agent has something to "discover."


**Project Suggestion 2: The "Climate-Aware Logistics Planner"**
**The Problem:** A shipping company needs to decide if they should reroute trucks based on weather and fuel costs.

* **Data Source:** * **Live API:** [OpenWeatherMap API](https://openweathermap.org/api) (Free Tier).
    * **Public Data:** A static CSV of "Major Shipping Routes" and their distances.
* **The Agentic Loop:**
    1.  User asks: "We have a shipment from Chicago to Denver leaving tomorrow. Should we take the I-80 or a southern route?"
    2.  **Tool 1 (Weather API):** Check the 48-hour forecast for major waypoints on both routes.
    3.  **Tool 2 (Python Math):** Calculate estimated fuel consumption based on distance, accounting for a "weather penalty" (e.g., +15% fuel for heavy snow/wind).
    4.  **Reasoning:** The agent weighs the cost of extra fuel vs. the risk of a "weather delay" to provide a recommendation.
* **Why it works:** It uses a real-world API, providing "live" stakes that feels like a real production app.


**Project Suggestion 3: The "GitHub Talent Scout" (Technical Analytics)**
**The Problem:** A tech recruiter needs to find open-source contributors who are "rising stars" in a specific niche (e.g., "Rust Web Frameworks").

* **Data Source:**
    * **Live API:** [GitHub REST API](https://docs.github.com/en/rest).
    * **RAG:** A PDF of the "Company Hiring Bar" (defining what a Senior vs. Junior dev looks like).
* **The Agentic Loop:**
    1.  User asks: "Find me 3 developers who are active in the `Axum` framework and have a high 'impact score'."
    2.  **Tool 1 (GitHub API):** Search for recent PRs and stars in specific repos.
    3.  **Tool 2 (Python/Reasoning):** Calculate an "Impact Score" based on (Commits * Stars) / Followers.
    4.  **Tool 3 (RAG):** Compare the top candidates against the "Company Hiring Bar" PDF to see if they fit.
* **Why it works:** It’s a very common real-world use case for agents. The GitHub API is well-documented and generous with its free tier.

## 12. Feasibility Checklist
When you propose your own project, considert the following three questions:
1.  **Can I query the data via a URL or a Local File?** (if you plan to scrape it, notice that not scraping is a distraction from the AI logic).
2.  **Is there a "Comparison" involved?** (Agents are best when choosing between Option A and Option B).
3.  **Is there a "Math Step"?** (To ensure it’s not just a chatbot, they *must* calculate something—a score, a cost, or a percentage).
